```python
#Athelet Participation vs Medal Performance 
#Counting Athelet by Country 
athelet_country = (
    athelets.groupby('NOC')
    .size()
    .reset_index(name = "Athelet_Count")
)
```

```python
athelet_country.head()
```

```python
#Merging it with Medal Data 
print(athelet_country['NOC'].head())
print(medals['Team/NOC'].head())
```

```python
#Checking the common countires before merging 
common_countries = set(athelet_country["NOC"]) & set(medals["Team/NOC"])

print("Matching countries:", len(common_countries))
```

```python
print(sorted(common_countries))
```

```python
#Checking unmatched countries
unmatched_athletes = sorted(
    set(athelet_country["NOC"]) - set(medals["Team/NOC"])
)

unmatched_medals = sorted(
    set(medals["Team/NOC"]) - set(athelet_country["NOC"])
)

print("Countries in athletes but not medals:")
print(unmatched_athletes)

print("\nCountries in medals but not athletes:")
print(unmatched_medals)
```

```python
country_mapping = {
    "People's Republic Of China": "People's Republic of China",
    "Republic Of Korea": "Republic of Korea",
    "Islamic Republic Of Iran": "Islamic Republic of Iran",
    "Côte D'Ivoire": "Côte d'Ivoire",
    "Republic Of Moldova": "Republic of Moldova",
    "United States Of America": "United States of America",
    "Roc": "ROC"
}

athelet_country["NOC"] = athelet_country["NOC"].replace(country_mapping)
```

```python
common_countries = set(athelet_country["NOC"]) & set(medals["Team/NOC"])

print("Matching countries:", len(common_countries))
```

```python
print("In athletes but not medals:")
print(sorted(
    set(athelet_country["NOC"]) - set(medals["Team/NOC"])
))

print("\nIn medals but not athletes:")
print(sorted(
    set(medals["Team/NOC"]) - set(athelet_country["NOC"])
))
```

```python
athelet_country["NOC"] = athelet_country["NOC"].replace(
    "People'S Republic Of China",
    "People's Republic of China"
)
```

```python
common_countries = set(athelet_country["NOC"]) & set(medals["Team/NOC"])

print("Matching countries:", len(common_countries))
```

```python
#Merging the countires
country_performance = athelet_country.merge(
    medals, 
    left_on = "NOC",
    right_on = "Team/NOC",
    how = "inner"
)
```

```python
country_performance.shape
```

```python
#Medal Efficiency 
country_performance['Medals_Per_100_Athelet'] = (
    country_performance['Total'] / country_performance['Athelet_Count'] * 100 
)
```

```python
#Top 10 Values 
top_efficiency = country_performance.sort_values(
    "Medals_Per_100_Athelet",
    ascending = False 
).head(10)
top_efficiency[[
    "NOC",
    "Athelet_Count",
    "Gold",
    "Silver",
    "Bronze",
    "Total",
    "Medals_Per_100_Athelet"
]]
```

```python
#Visualization 
plt.figure(figsize=(10, 6))

plt.barh(
    top_efficiency["NOC"],
    top_efficiency["Medals_Per_100_Athelet"]
)

font1 = {'family' : 'serif' , 'color' : 'darkblue' , 'size' : 20}
font2 = {'family' : 'serif' , 'color' : 'darkred' , 'size' : 12}

plt.xlabel("Medals per 100 Athletes", fontdict = font2)
plt.ylabel("Country", fontdict = font2)
plt.title("Top 10 Countries by Medal Efficiency", fontdict = font1)
plt.gca().invert_yaxis()

plt.show()
```

```python
#EDA questions and Insights 
'''
Question : Which countries have the highest medals per 100 athletes?
Insight : San Marino has the highest medals per 100 athletes.Therefore, smaller delegations can sometimes achieve high medal efficiency.
'''
```

```python
#Athelet Count vs Total Medals
plt.figure(figsize = (10,6))

plt.scatter(
    country_performance['Athelet_Count'],
    country_performance['Total']
)

font1 = {'family' : 'serif' , 'color' : 'darkblue' , 'size' : 20}
font2 = {'family' : 'serif' , 'color' : 'darkred' , 'size' : 12}

plt.xlabel("Number of Athletes", fontdict = font2)
plt.ylabel("Total Medals", fontdict = font2)
plt.title("Athlete Participation vs Total Medals", fontdict = font1)

plt.show()
```

```python
#EDA questions and Insights 
'''
Question : How does medal composition differ between countries?
Insight : Here the points generally move upward from left to right it means that the larger deligation tend to win more medals.
'''
```

```python
#Quantifying the relationship 
correlation = country_performance['Athelet_Count'].corr(country_performance['Total'])
print("Correlation :", correlation)
```

```python
'''
The correlation value is 0.875 which thereby means that there is strong relationship between the Number of Athelets and 
Total Medals among the medal-winning NOCs in the dataset.
'''
```
